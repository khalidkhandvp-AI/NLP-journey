***1_import re and writing corpus***

```python
import re
```

```python
corpus="My name is Khalid Khan. My mobile number is 03329142332 . My email is khalidkhan.dvp@gmail.com."
```

***2_Now lets find mobile number and email***

```python

email=re.findall(r'[\w.-]+@[\w.-]+\.\w+',corpus)
print(email)
```

```python
numb=re.findall(r'03\d{9}',corpus)
print(numb)
```
