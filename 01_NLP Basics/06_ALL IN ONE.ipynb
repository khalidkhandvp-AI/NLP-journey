**1_importing spacy and loading EN_CORE_WEB_SM**

```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
```

**2_writing text and load it to nlp**

```python
text="My name is Khalid Khan and I am from Buner"
```

```python
doc=nlp(text)
```

**3_pos tagging**

```python
for token in doc:
    print(token.text,"--->",token.pos_)
```

**4_tokenization**

```python
for token in doc:
    print(token.text)
```

**5_lemmatization**

```python
for token in doc:
    print(token.text,"-->",token.lemma_)
```

**6_stop words**

```python
for token in doc:
    print(token.text,"-->",token.is_stop)
```

**7_NER**

```python
for token in doc.ents:
    print(token.text,"--->",token.label_)
```
