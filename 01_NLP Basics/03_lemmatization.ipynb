**1_Import spacy , english model and writing text**

```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
text="I am studying AI from 2 years."
```

**2_Now send the senteence to sapcy for further lammatization**

```python
doc=nlp(text)
```

```python
for token in doc:
    print(token.text,"--->",token.lemma_)
```
