**1_Importing spacy and load english model**

```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
print("Model Loaded")
```

**2_Writing text**

```python
text="A quick brown fox jumps over the lazy dog."
```

**3_Process the text**

```python
doc=nlp(text)
for token in doc:
    print(token.text,token.pos_)
```

```python
text2="khalid studies AI"
doc1=nlp(text2)
for token in doc1:
    print(token.text,token.pos_)
```
