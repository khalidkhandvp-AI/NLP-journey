**1_ Import spacy, wriring text, load it to nlp**

```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
text="Elon Musk works at Microsoft in New York"
```

```python
doc=nlp(text)
```

**2_Apply loop and doing NER**

```python
for token in doc.ents:
    print(token.text,"--->",token.label_)
```
