**1_ Import spacy, wriring text, load it to nlp**

```python
import spacy
```

```python
nlp=spacy.load('en_core_web_sm')
```

```python
text="He is going to the gym right now"
```

```python
doc=nlp(text)
```

**2_applying the loop and finding the stop words**

```python
for token in doc:
    print(token.text,token.is_stop)
```
